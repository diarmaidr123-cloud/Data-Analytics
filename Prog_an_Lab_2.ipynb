```python
password = input('password')

def check_password():
    if len(password)>=8 and  any(char.isdigit()for char in password) and any(not char.isalnum()for char in password):
        print('Password is valid')
    else:
        print('Password is not valid')

check_password()
        
```

```python
def convert_temp(): 
    for Celsius in (range(-10,100)):
        Fahrenheit = (Celsius *9/5) +32
        print(Celsius, Fahrenheit)

convert_temp()    

    
```

```python
number=int(input())

def check_factorial():
    if number<0:
        print('Error')
    else:
        factor=1
        for x in range(1,number + 1):
            factor *=x

            print(factor)
        

check_factorial()
```

```python
number=int(input())

def prime_number():
    if number %2==0:
        print('Not prime')
    else:
        for x in range(2, number):
            if number % x == 0:
                print('Not prime')
        else:
            print('prime')

prime_number()
```
