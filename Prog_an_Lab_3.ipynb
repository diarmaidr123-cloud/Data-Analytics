Create a list with 3 student dictionaries	

```python
students ={
    1: { 'Name' : 'David', 'Age' : ' 23' , 'Grade' : '65' } ,
    2: { 'Name' : 'Maria', 'Age' : ' 26' , 'Grade' : '72' } ,
    3: { 'Name' : 'Chris', 'Age' : ' 29' , 'Grade' : '85' } }
```

Print the average grade

```python
average = sum(int(students[i]['Grade']) for i in students) / len(students)

print('Average:', average)
```

Print the names of the students who scored above average

```python
for i in students:
    if int(students[i]['Grade']) > average:
        print(students[i]['Name'])
```

Convert all names to upercase and print them

```python
for i in students:
    print(students[i]['Name'].upper())
```

Ask	the	user for a student	name and use try/except	to	check if it	exists in the	list.Print	
their grade	or an error message

```python
try:
    name = input('Name: ')
    
    for i in students:
        if students[i]['Name'] == name:
            print('Grade: ', students[i]['Grade'])
            break

    else: raise ValueError

except ValueError:
    print('Student not in class. Make sure first letter of name is a capital')
    


                
        
```
